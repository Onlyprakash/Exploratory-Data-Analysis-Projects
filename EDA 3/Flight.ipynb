```python
## Feature Engineering
## 1st--> Date of Journey

df['Date']=
```

```python
df['Date_of_Journey'].str.split('/').str[0]
## You can see the all date object into Integer
```

```python
df['Date']= df['Date_of_Journey'].str.split('/').str[0]
df['Month']= df['Date_of_Journey'].str.split('/').str[1]
df['Year']= df['Date_of_Journey'].str.split('/').str[2]
```

```python
df.head()
```

```python
df.info()
```

You can see [Date], [Month], [Year] are have individual column their but into again object then try into INT format astype(int) using Feature 

```python
df['Date']=df['Date'].astype(int)
df['Month']=df['Month'].astype(int)
df['Year']=df['Year'].astype(int)
```

```python
df.info()
```

check again and you can see the [Date],[Month],[Year] are those into Interger format and we don't need to [Date of Journey] column

```python
## Drop Date of Journey column

df.drop('Date_of_Journey',axis=1,inplace=True)
```

```python

df.head()
```

```python
df['Arrival_Time']=df['Arrival_Time'].apply(lambda x: x.split(' ')[0])
```

```python
## Feature Engineering
## 2nd--> Arrival Time

df['Arrival_hour']=df['Arrival_Time'].str.split(':').str[0]
df['Arrival_min']=df['Arrival_Time'].str.split(':').str[1]
```

```python
df.head(2)
```

```python
df['Arrival_hour']=df['Arrival_hour'].astype(int)
df['Arrival_min']=df['Arrival_min'].astype(int)
```

```python
df.head()
```

```python
## Feature Engineering
## 3rd--> Depature Time

df['Depature_Hour']=df['Dep_Time'].str.split(':').str[0]
df['Depature_min']=df['Dep_Time'].str.split(':').str[1]
```

```python
df.head()
```

```python
df.info()
```

```python
df['Depature_Hour']=df['Depature_Hour'].astype(int)
df['Depature_min']=df['Depature_min'].astype(int)
```

```python
df.info()
```

```python
df.drop('Dep_Time',axis=1,inplace=True)
```

```python
df.head(2)
```

```python
## Feature Engineering
## 4th--> Total Stops

df['Total_Stops'].unique()
```

```python
df['Total_Stops'].isnull()
```

```python
df['Total_Stops']=df['Total_Stops'].map({'non-stop':0,'1 stop':1,'2 stops':2,'3 stops':3,'4 stops':4,np.nan:1})
```

```python
df[df['Total_Stops'].isnull()]
```

```python
df.head()
```

```python
df.drop('Route',axis=1,inplace=True)
```

```python
df.head(2)
```

```python
## Feature Engineering
# Duration

df['Duration'].str.split(' ').str[0]
```

```python
df['Duration'].str.split(' ').str[0].str.split('h').str[0]
```

```python
df['Duration'].str.split(' ').str[1].str.split('h').str[0]
```

```python
df['Duration'].str.split(' ').str[1].str.split('m').str[0]
```

```python
df['Duration_Hour']=df['Duration'].str.split('h').str[0]
df['Duration_min']=df['Duration'].str.split('m').str[1]
```

```python
df.head(10)
```

```python
df.info()
```

```python
df['Duration_Hour'] = df['Duration'].str.extract(r'(\d+)h').fillna(0).astype(int)
df['Duration_min'] = df['Duration'].str.extract(r'(\d+)m').fillna(0).astype(int)
```

```python
df.info()
```

```python
df.head(10)
```

```python
df.drop('Duration',axis=1,inplace=True)
```
