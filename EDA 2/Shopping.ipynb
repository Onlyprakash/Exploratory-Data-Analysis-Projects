```python
## Import Libaries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
## Read the Dataset

df= pd.read_csv('shopping_trends.csv')
```

```python
## Data Visualization
# Histogram

df["Age"].plot(kind = "hist")
plt.title("Histogram of Age")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

```python
df.columns
```

```python
## Barplot 

df["Gender"].value_counts().plot(kind="bar")
plt.title("Barplot of Gender Distribution")
plt.xlabel("Gender")
plt.ylabel("Count") 
plt.show()
```

```python
df.groupby('Category')['Purchase Amount (USD)'].mean()
```

```python
df.sort_values('Review Rating', ascending=False).head(3)
```

```python
df.sort_values('Review Rating', ascending=False, inplace=True)
```

```python
## Pie chart 

df['Subscription Status'].value_counts().plot(kind= 'pie')
plt.title('Pie Chart of Subscription Status')
plt.show()
```

```python
df['Color'].value_counts().head(10)
```

```python
df['Color'].value_counts().head(10).plot(kind= 'pie')
```

```python
## Scatter plot 

df.head(15).plot(x='Previous Purchases', y='Review Rating', kind='scatter')
plt.title('Scatter Plot of Preview Purchase vs Review Rating')
plt.xlabel('Preview Purchase')
plt.ylabel('Review Rating')
plt.show()
```

```python
df.groupby('Season')['Purchase Amount (USD)'].sum()
```

```python
## BoxPlot

df.boxplot(column = 'Purchase Amount (USD)', by = 'Frequency of Purchases')
plt.title('Box Plot of Purchase Amount by Frequency of Purchases')
plt.show()
```

```python
## Barchart

df['Season'].value_counts().plot(kind='bar')
plt.xlabel('Season')
plt.ylabel('Count ')
plt.title('Count of Purchases by Season')
plt.show()
```

```python
## Pie chart

df['Category'].value_counts().plot(kind='pie')
plt.title('Distribution of Purchases by Category')
plt.show()
```

```python
## Boxplot

df.boxplot(column='Review Rating', by='Gender')
plt.xlabel('Gender')
plt.ylabel('Review Rating ')
plt.title('Review Rating Distribution by Gender')
plt.show()
```

```python
## Histogram

df['Purchase Amount (USD)'].plot(kind='hist', bins=10)
plt.xlabel('USD')
plt.ylabel('Frequency')
plt.title('Distribution of Purchase Amounts')
plt.show()
```

```python
## Barchart

df.groupby('Color')['Review Rating'].mean().plot(kind='bar')
plt.xlabel('Color')
plt.ylabel('Mean Review Rating')
plt.title('Mean Review Rating by Color')
plt.show()
```

```python
## pie chart

df.groupby('Payment Method')['Purchase Amount (USD)'].sum().head().plot(kind='pie')
plt.title('Sum of Purchase Amount by Payment Method')
plt.ylabel('')  # Hide y-label for better appearance
plt.show()

```
